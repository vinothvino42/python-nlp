# One Hot Encoding

## Without using Libraries

```python
categories = ["red", "blue", "green", "blue", "red"]

# Get unique values
unique_values = list(set(categories))

# Create one-hot encoding
one_hot_encoded = []
for item in categories:
    row = [1 if item == val else 0 for val in unique_values]
    one_hot_encoded.append(row)

print("Categories:", categories)
print("Unique values:", unique_values)
print("One-hot encoded result:")
for row in one_hot_encoded:
    print(row)
```

## Using Pandas

```python
import pandas as pd

# Sample data
data = {"color": ["red", "blue", "green", "blue", "red"]}

# Create DataFrame
df = pd.DataFrame(data)

# One-hot encode
one_hot = pd.get_dummies(df, columns=["color"])

print("Original Data:")
print(df)

print("\nOne-Hot Encoded Data:")
print(one_hot)
```
